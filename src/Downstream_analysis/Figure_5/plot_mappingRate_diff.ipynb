```python
import pandas as pd
import matplotlib.pyplot as plt
```

From the mapping_summary generated with the E-mapping_rates.sh script in HPC_analysis/4-quantification, you can run this Python file. 

Specifically, it's designed to compare the mapping efficiencies between gencode and refseq.

```python
k15 = pd.read_table('mapping_summary_gencode_k15.txt', sep='\t')
k15_refseq = pd.read_table('mapping_summary_refseq_k15.txt', sep='\t')
rRNA = pd.read_table('mapping_EMN_45S_rRNA.tsv', sep='\t')
```

```python
k15['MappingRate_float'] = k15['MappingRate'].str.rstrip('%').astype(float)
k15_refseq['MappingRate_float'] = k15_refseq['MappingRate'].str.rstrip('%').astype(float)
rRNA['MappingRate_float'] = rRNA['MappingRate'].str.rstrip('%').astype(float)

# Ensure samples are matched and in the same order
merged = pd.merge(k15[['Samples', 'MappingRate_float']], 
				  k15_refseq[['Samples', 'MappingRate_float']], 
				  on='Samples', 
				  suffixes=('_gencode', '_refseq'))

merged2 = pd.merge(merged, rRNA[['Samples', 'MappingRate_float']], on='Samples', suffixes=('', '_rRNA'))

merged2 = merged2.sort_values(by='Samples').reset_index(drop=True)
```

```python
merged2 = merged2[~merged2['Samples'].str.contains('DC')]
```

```python
plt.figure(figsize=(12, 6))
plt.plot(merged2['Samples'], merged2['MappingRate_float_gencode'], marker='o', label='GENCODE', color='cornflowerblue')
plt.plot(merged2['Samples'], merged2['MappingRate_float_refseq'], marker='o', label='RefSeq', color='limegreen')
#plt.plot(merged2['Samples'], merged2['MappingRate_float'], marker='o', label='index_k15_rRNA', color='darkorange')
plt.xticks(rotation=90)
plt.xticks([], [])
plt.yticks(size=12)
plt.ylabel('Mapping Rate (%)', size=12)
plt.xlabel('Samples', size=12)
plt.legend()
plt.tight_layout()
plt.show()
```
